````python
from pathlib import Path
from datetime import datetime

import matplotlib.pyplot as plt
import mlflow
from mlflow.tracking import MlflowClient
import pandas as pd
from sklearn.metrics import accuracy_score, f1_score, precision_score, recall_score, roc_auc_score

# Add imports from your selected exercise as needed.
# Example:
# from sklearn.linear_model import LogisticRegression
# from xgboost import XGBClassifier
````

````python
def find_project_root(start: Path) -> Path:
    for path in [start, *start.parents]:
        if (path / "data").exists() and (path / "notebooks").exists():
            return path
    raise FileNotFoundError("Could not find project root. Please run this notebook inside the project repository.")

PROJECT_ROOT = find_project_root(Path.cwd())
MLFLOW_DB = PROJECT_ROOT / "mlflow.db"
MLRUNS_DIR = PROJECT_ROOT / "mlruns"
EXERCISE_RESULTS_DIR = PROJECT_ROOT / "reports" / "exercise_results"

MLRUNS_DIR.mkdir(parents=True, exist_ok=True)
EXERCISE_RESULTS_DIR.mkdir(parents=True, exist_ok=True)

print("Project root:", PROJECT_ROOT)
print("MLflow database:", MLFLOW_DB)
````

## 4. Configure MLflow

Use a separate experiment name for exercise work. This keeps the main Telco Churn runs separate from your own practice runs.

````python
tracking_uri = f"sqlite:///{MLFLOW_DB}"
experiment_name = "exercise-mlflow-tracking"
artifact_location = (MLRUNS_DIR / experiment_name).as_uri()

mlflow.set_tracking_uri(tracking_uri)
client = MlflowClient(tracking_uri=tracking_uri)

experiment = client.get_experiment_by_name(experiment_name)
if experiment is None:
    experiment_id = client.create_experiment(
        name=experiment_name,
        artifact_location=artifact_location,
    )
else:
    experiment_id = experiment.experiment_id

mlflow.set_experiment(experiment_name)

print("Tracking URI:", mlflow.get_tracking_uri())
print("Experiment ID:", experiment_id)
print("Artifact location:", artifact_location)
````
