# 06 Exercise: Understand Docker Compose and CI

In this exercise, you do not build a new deployment pipeline from scratch. The goal is to understand the local container setup and the GitHub Actions workflow used in this project.

Docker and CI/CD introduce several new concepts at once. Therefore, this exercise focuses on reading, running, and explaining the existing setup.

## 1. Start the Local Application with Docker Compose

From the repository root, start the local application:

```bash
docker compose up --build
```

Open the two local interfaces:

```text
FastAPI Swagger UI: http://127.0.0.1:8000/docs
Streamlit app:      http://127.0.0.1:8501
```

Stop the stack when you are done:

```bash
docker compose down
```
