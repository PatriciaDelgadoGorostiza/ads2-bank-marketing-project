## 4. Write a Small Unit Test

For each extracted function, write at least one small test.

Use artificial mini data where possible. A test should make the expected behavior obvious.

Example:

```python
def test_create_tree_age_features_calculates_age():
    df = pd.DataFrame({"pflanzjahr": [2000]})

    result = create_tree_age_features(df, reference_year=2025)

    assert result.loc[0, "tree_age"] == 25
```

If setup data is reused across tests, move it into a fixture in `tests/conftest.py`.

## 5. Unit Test or Integration Test?

Classify your test:

- **Unit test:** one function, small artificial input, fast, no file dependencies.
- **Integration test:** multiple components together, for example loading data, transforming it, and running a model or pipeline step.

For this exercise, start with unit tests. Add an integration test only if it is still small and fast.

## 6. Run Your Tests

From the repository root:

```bash
python -m pytest
```

If you use markers:

```bash
python -m pytest -m unit
python -m pytest -m integration
```

A useful first goal is not high coverage. A useful first goal is one or two tests that protect important project assumptions.
